```python
import pandas as pd 
```

```python
file_name = "car_sales.csv"
df = pd.read_csv(file_name)
```

```python
df.head(3)
```

```python
df.value_counts("Company")
```

```python
df.nunique()
```

```python
#Honda araclarinin ortalama satis fiyati
df.query("Company == 'Honda'")["Selling_Price"].mean()
```

```python
import matplotlib.pyplot as plt
```

## 1. Markalara göre ortalama fiyat

```python
df.groupby("Company")["Selling_Price"].mean().sort_values().plot(kind="barh")

plt.title("Markalara Göre Ortalama Satış Fiyatı")
plt.xlabel("Ortalama Fiyat")
plt.ylabel("Marka")
plt.show()
```

## 2.Yakıt tipine göre araç sayısı

```python
df["Fuel_Type"].value_counts().plot(kind="bar")

plt.title("Yakıt Tipine Göre Araç Sayısı")
plt.xlabel("Yakıt Tipi")
plt.ylabel("Araç Sayısı")
plt.show()
```

## 3.Vites tipine göre dağılım

```python
df["Transmission"].value_counts().plot(kind="pie", autopct="%1.1f%%")

plt.title("Vites Tipi Dağılımı")
plt.ylabel("")
plt.show()
```

## 4. Yıl ve satış fiyatı ilişkisi

```python
plt.scatter(df["Year"], df["Selling_Price"])

plt.title("Yıla Göre Satış Fiyatı")
plt.xlabel("Yıl")
plt.ylabel("Satış Fiyatı")
plt.show()
```

## 5. Motor hacmi ve fiyat ilişkisi

```python
plt.scatter(df["Engine_CC"], df["Selling_Price"])

plt.title("Motor Hacmi ve Satış Fiyatı İlişkisi")
plt.xlabel("Motor CC")
plt.ylabel("Satış Fiyatı")
plt.show()
```
